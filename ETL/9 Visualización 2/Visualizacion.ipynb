```python
ganancias_ventas.index.name = 'Tipo de producto'
ganancias_ventas.head()
```

```python
ganancias_ventas_style = ganancias_ventas.style
```

```python
ganancias_ventas_style = ganancias_ventas_style.format ('USD {:,.2f}'). \
                                                highlight_max (color= 'lightgreen'). \
                                                highlight_min (color= 'red') #highlight es resaltar

ganancias_ventas_style
```

## *Pedidos por region*

```python
pedidos_region = ventas['region'].value_counts().to_frame()
pedidos_region
```

```python
pedidos_region.index.name = 'Región'
pedidos_region.rename (columns= {'count': 'N° pedidos'}, inplace= True)
porcentaje = pedidos_region['N° pedidos'].to_numpy()
porcentaje = porcentaje*100 / np.sum(porcentaje)
porcentaje
```

```python
pedidos_region ['Ponderación de pedidos'] = porcentaje
estilo_region = pedidos_region.style
```

```python
estilo_region
```

```python
estilo_region.format ({'Ponderación de pedidos': '{:.2f} %'}) \
                    .bar(subset= 'Ponderación de pedidos', vmin= 0, vmax= 100, cmap='Reds')

# se puede utilizar color = '#COLOR' en vez de cmap...

# 'Greens' - verde claro a oscuro
# 'Blues' - azul claro a oscuro
# 'Reds' - rojo claro a oscuro
# 'Oranges' - naranja claro a oscuro
# 'YlGn' - amarillo a verde
# 'GnBu' - verde a azul
# 'PuBu' - púrpura a azul
```

## *Formatos DataTime*

```python
meses = ventas.copy()
```

```python
meses = meses.sort_values ('fecha_pedido')
meses ['mes'] = meses ['fecha_pedido'].dt.strftime ('%Y - %B') # El simbolo %B significa MES en MAYÚSCULAS
meses.reset_index (drop= True, inplace= True)
meses
```

### *Tabla dinámica de resumen*

- *Pivot Table lo que hace es hacer un resumen con estadísticas básicas en una tabla (como sumas, restas, divisiones, promedios, etc)*
- *Pivot en su defecto, toma todos los datos y los separa con los parámetros que uno desee, pudiendo agruparlos de forma más eficiente o separarlos en distintas categorías*

```python
ventas_mensuales = meses.pivot_table (index= meses.departamento, columns= 'mes', values= 'cantidad', aggfunc= 'sum', sort= False) # False por defecto
ventas_mensuales
```

```python
pd.set_option ('display.max_columns', None) # Para ver todas las columnas
ventas_mensuales.head()
# Volver a resumir las columnas por default: pd.reset_option('display.max_columns')
```

```python
estilo_mensual = ventas_mensuales.style
estilo_mensual.set_sticky (axis = 'index') # Para fijar los índices
```
