```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_json ('datos/datos_hosting.json')
data = pd.json_normalize (data['info_inmuebles'])
```

```python
data.head()
```

# **Tokenización descripción_vecindad**

```python
data['descripcion_vecindad'] = data['descripcion_vecindad'].fillna("").astype(str).str.lower().str.strip()
```

```python
data['descripcion_vecindad'] = data['descripcion_vecindad'].str.replace(r'[\[\]\"\']', '', regex=True)
data['descripcion_vecindad'] = data['descripcion_vecindad'].str.replace('[^a-zA-z0-9\-\']', ' ', regex= True)
data['descripcion_vecindad'] = data['descripcion_vecindad'].str.replace('(?<!\w)-(?!\w)', '', regex=True)
```

```python
data['descripcion_vecindad'] = data['descripcion_vecindad'].str.replace('[{}"]', '', regex = True)
```

```python
data['descripcion_vecindad'] = data['descripcion_vecindad'].str.split()
```
