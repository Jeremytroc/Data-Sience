```python
data [['cuota_deposito', 'cuota_limpieza']] = data [['cuota_deposito', 'cuota_limpieza']].map(lambda x: x.replace ('$' , '').replace (',', '').strip())
```

```python
data [['cuota_deposito', 'cuota_limpieza']] = data [['cuota_deposito', 'cuota_limpieza']].astype (np.float64)

data.info()
```

# **Tokenización**

---
- *En la tokenización lo que se busca es tomar elementos strings de una serie (comentarios, opiniones, descripción de un producto, etc) y buscar keywords para asignar una ponderación de importancia u otro factor relevante a esta para que explique a una variable dependiente*

## *Limpieza de serie*

- *Para este ejercicio se utilizará la serie "descripcion_local" para limpiarla y confeccionar posteriormente el algoritmo*

```python
data['descripcion_local'] = data['descripcion_local'].str.lower()
data.info()
```

*Para eliminar símbolos extraños se utiliza: **[^a-zA-Z0-9\-\']** y **(?<!\w)-(?!\w)***

- "^" sirve para negación en un reemplazo, es decir: no reemplazar lo siguiente, pero sí todo lo demás
- a-z, A-Z para referirse a cualquier letra/palabra y 0-9 para cualquier número
- \- y \' para que no reemplace las comillas simples ni el guión
---
- Podría existir una palabra como "two - bedroom" y la idea es tokenizar la frase completa, no separar todo, eso se hace con (?<!\w)-(?!\w)

```python
data['descripcion_local'] = data['descripcion_local'].str.replace('[^a-zA-Z0-9\-\']',' ', regex=True)
data['descripcion_local'] = data['descripcion_local'].str.replace('(?<!\w)-(?!\w)',' ', regex=True)
```
