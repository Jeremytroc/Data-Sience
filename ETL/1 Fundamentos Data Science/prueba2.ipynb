```python
'''Crea un programa que tenga la siguiente lista con los gastos de una empresa de papel 
[2172.54, 3701.35, 3518.09, 3456.61, 3249.38, 2840.82, 3891.45, 3075.26, 2317.64, 3219.08].
Con estos valores, crea un programa que calcule el promedio de gastos. Sugerencia: usa las funciones integradas sum() y len().
'''
import math
gastos = [2172.54, 3701.35, 3518.09, 3456.61, 3249.38, 2840.82, 3891.45, 3075.26, 2317.64, 3219.08]
promedio = sum(gastos)/len(gastos)
print (promedio)
```

```python
'''Con los mismos datos de la pregunta anterior, determina cuántas compras se realizaron por encima de 3000 reales 
y calcula el porcentaje con respecto al total de compras.
'''
suma_total = len(gastos)
print (suma_total)
contador_gasto = 0

for gasto in gastos:
    if gasto > 3000:
        contador_gasto += 1


porcentaje = contador_gasto/suma_total*100
print(porcentaje)
```

```python
#Crea un programa que, al ingresar un número cualquiera, genere una lista que contenga todos los números primos entre 1 y el número ingresado.
lista = []
for i in range(1,6):
    numero = int(input(f'{i}. Ingrese un numero: '))
    lista.append(numero)

for n in range(2, max(lista) + 1):
    # asumimos que n es primo hasta que encontremos un divisor
    es_primo = True
    for i in range(2, n):
        if n % i == 0:
            es_primo = False
            break  # deja de revisar si encontramos un divisor
    if es_primo:
        print(n, end=' ')
```

```python
'''Para un estudio sobre la multiplicación de bacterias en una colonia, se recopiló el número de bacterias multiplicadas por día y
se puede observar a continuación: [1.2, 2.1, 3.3, 5.0, 7.8, 11.3, 16.6, 25.1, 37.8, 56.9].
Con estos valores, crea un código que genere una lista que contenga el porcentaje de crecimiento de bacterias por día,
comparando el número de bacterias en cada día con el número de bacterias del día anterior. Sugerencia: para calcular el porcentaje de crecimiento, 
utiliza la siguiente ecuación: 100 * (muestra_actual - muestra_anterior) / muestra_anterior.
'''
lista = [1.2, 2.1, 3.3, 5.0, 7.8, 11.3, 16.6, 25.1, 37.8, 56.9]
crecimiento = []

for elemento in range (1,len(lista)):
    tasa = (lista[elemento]/lista[elemento-1]-1)*100
    crecimiento.append(round(tasa,2))


print (f'Los porcentajes de incremento son los siguientes: \n{crecimiento}')
```

```python
'''Para una selección de productos alimenticios, debemos separar el conjunto de IDs proporcionados por números enteros,
sabiendo que los productos con ID par son dulces y los que tienen ID impar son amargos.
Crea un código que recoja 10 IDs. Luego, calcula y muestra la cantidad de productos dulces y amargos.
'''
productos = []

for i in range (1,11):
    productos.append(int(input(f'Por favor ingrese el {i}° ID de producto: ')))

dulces = 0
amargos = 0

for id in productos:
    if id % 2 == 0:
        dulces += 1
    else:
        amargos += 1

print (f'Los productos dulces son: {dulces}')
print (f'\nLos productos amargos son: {amargos} de un total de: {dulces+amargos} productos')

```

```python
'''
Ana está implementando un sistema de filtrado de libros en Buscante. 
La funcionalidad debe recorrer una lista de libros y mostrar el nombre de cada libro disponible en stock. 
Sin embargo, si el libro está agotado, debe ser ignorado durante la iteración.
'''
libros = [
    {"nombre": "1984", "stock": 5},
    {"nombre": "Dom Casmurro", "stock": 0},
    {"nombre": "El Principito", "stock": 3},
    {"nombre": "El Hobbit", "stock": 0},
    {"nombre": "Orgullo y Prejuicio", "stock": 2}
]

for libro in libros:
    if libro["stock"] == 0:
        continue  # salta los agotados
    else:
        print(f'Libro disponible: {libro["nombre"]}')
```

```python
'''
Has recibido un desafío para calcular la raíz cuadrada de una lista de números,
identificando cuáles resultan en un número entero. La lista es la siguiente:
'''
import math
numeros = [2, 8, 15, 23, 91, 112, 256]
raices = []

for raiz in numeros:
    if math.sqrt(raiz).is_integer():
        raices.append(math.sqrt(raiz))
    else:
        continue

print (f'Las raices calculadas enteras son las siguientes: \n{raices}')
    

```
