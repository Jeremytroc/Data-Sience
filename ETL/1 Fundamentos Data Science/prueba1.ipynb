***2. Escribe un programa que solicite el porcentaje de crecimiento de producción de una empresa e informe si hubo un crecimiento (porcentaje positivo) o una disminución (porcentaje negativo)***.

```python
crecimiento = input('Por favor diga el % de crecimiento de su empresa').strip()
crecimiento = float(crecimiento.strip('%'))

if crecimiento >= 0:
    print ('Perfecto, el crecimiento de su empresa es de un {}%'.format(crecimiento))
else:
    print ('Su empresa no ha crecido')
```

```python
# Tabla del  7:
# 7 x 0 = 0
# 7 x 1 = 7
# [...]
# 7 x 10 = 70
resultado = []
def tablas(numero):
    for i in range (1,11):
        lista = numero*i
        resultado.append(lista)
        if i == 10:
            print (f'La tabla de multiplicar del {numero} es la siguiente: \n {resultado}')
        else:
            continue

tablas(2)

#OTRA FORMA
def multiplicar (valor):
    final_resultado = list(map(lambda x: x * valor, range(1,11)))
    print (f'La tabla de multiplicar del {valor} es la siguiente: \n {final_resultado}')
    return final_resultado

multiplicar(4)
resultado_final = multiplicar(4)
print (f'Vamos a ver cual es el valor del segundo resultado de la tabla realizada: {resultado_final[2-1]}.')

```

```python
nombres = ["juan", "MaRia", "JOSÉ"] 
apellidos = ["SILVA", "sosa", "Tavares"]
import numpy as np 
# Normalizar nombres y apellidos y crear una nueva lista con los nombres completos # Puedes apoyarte en la función map() 
nombres_normalizados = list(map(lambda x: x.capitalize(), nombres)) 
apellidos_normalizados = list(map(lambda x: x.capitalize(), apellidos)) 

nombres_completos = list(map(lambda x,y: f'nombre completo: {x} {y}', nombres_normalizados,apellidos_normalizados)) 

print (nombres_completos) 
print (f'probemos un vector :)\n') 

vector = np.column_stack ((nombres_normalizados,apellidos_normalizados)) 
print (vector)
```

```python
'''Crea un diccionario usando la comprensión de diccionarios (dict comprehension) en el que las claves estén en la
lista meses = ['Jan', 'Fev', 'Mar', 'Abr', 'Mai', 'Jun', 'Jul', 'Ago', 'Set', 'Out', 'Nov', 'Dez']
y los valores estén en gasto = [860, 490, 1010, 780, 900, 630, 590, 770, 620, 560, 840, 360].
'''
meses = ['Jan', 'Fev', 'Mar', 'Abr', 'Mai', 'Jun', 'Jul', 'Ago', 'Set', 'Out', 'Nov', 'Dez']
gasto = [860, 490, 1010, 780, 900, 630, 590, 770, 620, 560, 840, 360]

diccionario = {mes: monto for mes,monto in zip(meses, gasto)}
print (diccionario)
# o también lo siguiente
diccionario_2 = dict(zip(meses,gasto))
print (diccionario_2)
```
