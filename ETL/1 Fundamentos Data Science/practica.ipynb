```python
#Escribe un programa que solicite dos números enteros e imprima todos los números enteros entre ellos.
numero1 = int(input('Por favor ingrese un numero'))
numero2 = int(input('Por favor ingrese otro numero'))

if numero1 < numero2:
    for i in range (numero1+1,numero2):
        print (f'la diferencias entre los numeros son: {i}')
elif numero1 > numero2:
    for i in range (numero2+1,numero1):
        print (f'la diferencia entre los numeros son: {i}')
else:
    print ('Los numeros son iguales')
```

```python
'''Escribe un programa para calcular cuántos días tomará que la colonia de una bacteria A supere o iguale a la colonia de una bacteria B,
basado en tasas de crecimiento del 3% y 1.5%, respectivamente.
Supón que la colonia A comienza con 4 elementos y B con 10.
'''
bacteriaA = 4
bacteriaB = 10
dias = 0

while bacteriaA <= bacteriaB and dias < 1000:
    bacteriaA *= 1.03
    bacteriaB *= 1.015
    dias += 1

print (f'La bacteria A superará a la B en {dias} días')
```

```python
#Escribe un programa que calcule el factorial de un número entero proporcionado por el usuario.

numero = int(input('Ingrese un numero por favor, sólo uno: '))

if numero > 0:
    factorial = 1
    for i in range (1, numero+1):
        factorial *= i
    
    print (f'El factorial de tu número {numero} es: {factorial}')
else:
    print ('No puede ser un número negativo')
```
