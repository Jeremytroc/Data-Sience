### **Regresión lineal**


```python
from sklearn.linear_model import LinearRegression
```

```python
fechas = np.array(fechas).reshape(-1,1)
Moscu = np.array(Moscu)

# Modelo
model = LinearRegression()
model.fit(fechas, Moscu)

# Recta estimada
pred = model.predict(fechas)

# Gráfico simple
plt.plot(fechas, Moscu)
plt.plot(fechas, pred)
plt.grid(True)
plt.show()

print (model.coef_)
print (model.intercept_)
```

### Intercepto y coeficientes del modelo

- **El modelo queda de la siguiente forma:** $$\hat{y} = 79.6489 + 0.4352 \cdot \text{fechas}$$


```python
# Gráfico simple
plt.plot(fechas, Moscu)
plt.plot(fechas, Kaliningrado) #Agregamos Kaliningrado sólo para ver la diferencia de precios
plt.plot(fechas, pred)
plt.grid(True)
plt.plot(25, 79.64892274792835+0.43520212*25, '*r') #Agregar puntito rojo en el mes 25
plt.legend (['Moscú', 'Kaliningrado'])
plt.show()

print (model.coef_)
print (model.intercept_)
```

- Es importante el uso de .predict, dado que con .fit, uno entrena al modelo... ya tiene los coeficientes (.coef_) y el intercepto (.intercept_)
- Lo anterior se da para que, al momento de hacer .predict(array...) uno pueda agregar los datos que uno quiera predecir en las variables independientes, por ej; si agregamos un array de los mismos valores de x, obtenemos la estimación en los datos conocidos, si agregamos nuevos valores, entonces podremos predecir nuevos datos al agregar un nuevo vector.

```python
model2 = LinearRegression()
model2.fit (fechas, np.array(data['Petersburgo']))
pred2 = model2.predict (fechas)

nuevas_fechas = np.arange(88,97).reshape(-1,1)
pred2_1 = model2.predict (nuevas_fechas)

plt.plot (fechas, data['Petersburgo'])
plt.plot (fechas, pred2)
plt.plot (88, model2.predict(np.array([[88]])), '*r')
plt.plot (nuevas_fechas, pred2_1)
plt.grid(True)
plt.show()
```
